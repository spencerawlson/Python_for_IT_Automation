Bank ATM – Python Exercise
 
Create a Python program that simulates a simple ATM.
 
Requirements:
 
Start with a bank balance of $1,000.

Ask the user what they want to do:

1 – Deposit money

2 – Withdraw money

3 – Check balance

4 – Exit

For a deposit, ask how much money they want to deposit and add it to the balance.

For a withdrawal, ask how much they want to withdraw.

If they have enough money, subtract it.

Otherwise, print "Insufficient funds!"

After every transaction, display the current balance.

Keep showing the ATM menu until the user chooses Exit.
 
Bonus: Don't allow negative deposits or withdrawals.
 

```python
balance = 1000.00

print(f'Your current balance is: ${balance:.2f}')

print(
    '1 - Deposit\n'
    '2 - Withdraw\n'
    '3 - Check Balance\n'
    '4 - Exit\n'
)

while True:

    option = int(input('What would you like to do? '))

    if option == 1:
        deposit = float(input('How much would you like to deposit? '))
        if deposit <= 0:
            print('Cannot deposit zero or a negative amount')
        else:
            balance = balance + deposit
            print(f'Your new balance is: ${balance:.2f}')
    elif option == 2:
        withdraw = float(input('How much would you like to withdraw? '))
        if withdraw <= 0:
            print('Cannot withdraw zero or a negative amount')
        elif withdraw > balance:
            print('Insufficient Funds')
        else:
            balance = balance - withdraw
            print(f'Your new balance is: ${balance:.2f}')
    elif option == 3:
        print(f'Your current balance is: ${balance:.2f}')
    elif option == 4:
        print('Thank you for using the banking system.')
        break
    else:
        print('Invalid option. Please choose between 1 and 4.')
```
